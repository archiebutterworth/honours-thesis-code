# Comparing noiseless, emulator and hardware shadow overlaps

Loads the three result sets and computes the mean shadow overlap for each:

| source | mean omega |
|---|---|
| noiseless simulation | 0.984 |
| H2-1 emulator | 0.971 |
| **H2-1 hardware** | **0.967 +/- 0.024** (standard error, 1000 shots) |

The emulator and hardware agreeing to within the statistical uncertainty of the
1000-shot estimates is the evidence that the emulator captures the device's
dominant noise processes, and therefore that the gap from unity is hardware noise
rather than a limitation of the preparation algorithm.

Runs offline -- no Quantinuum account required.


```python
import pickle

with open("shadow_results_hardware.pkl", "rb") as f:
    shadow_results_hardware = pickle.load(f)
```

```python
import pickle

with open("shadow_results_emulator.pkl", "rb") as f:
    shadow_results_emulator = pickle.load(f)
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = shadow_results_emulator["all results"]

mean = np.mean(data)

plt.figure(figsize=(8,5))

plt.hist(
    data,
    bins=50,
    color='grey',
    edgecolor='black',
    alpha=0.8
)

plt.axvline(
    mean,
    linestyle='--',
    linewidth=2,
    label=f"Mean = {mean:.3f}"
)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Shadow Overlap Results from H2-1 Emulator (1000 Shots)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
data = shadow_results_hardware["all results"]

mean = np.mean(data)

plt.figure(figsize=(8,5))

plt.hist(
    data,
    bins=50,
    color='grey',
    edgecolor='black',
    alpha=0.8
)

plt.axvline(
    mean,
    linestyle='--',
    linewidth=2,
    label=f"Mean = {mean:.3f}"
)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Shadow Overlap Results from H2-1 (1000 Shots)")
plt.legend()
plt.tight_layout()
plt.show()
```
